```python
class Solution:
    def count_digits(self, n):
        count = 0
        num = abs(n)
        if num == 0:
            return 1
        while num > 0:
            count += 1
            num //= 10
        return count

sol = Solution()

while True:
    try:
        n = int(input("Enter a number: "))
        break
    except ValueError:
        print("Invalid input! Please enter a valid integer.")

print(f"Number of digits: {sol.count_digits(n)}")
```

```python
class Solution:
    def check_palindrome(self, s):
        num = abs(s)
        result = 0
        while num > 0:
            digit = num % 10
            result = result * 10 + digit
            num //= 10
        return result == abs(s)
    
sol = Solution()
n = int(input("enter a number:"))

if sol.check_palindrome(n):
        print(f"{n} is a palindrome.")
else:
        print(f"{n} is not a palindrome.")
```

```python
class Solution(object):
    def search(self, nums, target):
        left = 0
        right = len(nums) - 1
        
        while left <= right:
            mid = left + (right - left) // 2
            
            if nums[mid] == target:
                return mid
            
            # Check if the left half is sorted
            if nums[left] <= nums[mid]:
                # Check if target is within the sorted left half
                if target >= nums[left] and target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
                    
            # Otherwise, the right half must be sorted
            else:
                # Check if target is within the sorted right half
                if target > nums[mid] and target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
                    
        # Target was not found
        return -6
sol = Solution()
nums = input("Enter a sorted array (comma-separated): ")
nums = [int(x.strip()) for x in nums.split(',')]
target = int(input("Enter the target value: "))
result = sol.search(nums, target)
if result != -6:
    print(f"Target {target} found at index: {result}")
else:
    print(f"Target {target} not found in the array.")
```

```python
class Solution:
    def armstromng(self,n):
        num = abs(n)
        sum_of_digits = 0
        num_digits = len(str(num))
        
        while num > 0:
            digit = num % 10
            sum_of_digits += digit ** num_digits
            num //= 10
            
        return sum_of_digits == abs(n)
sol = Solution()
n = int(input("Enter a number: "))
if sol.armstromng(n):
    print(f"{n} is an Armstrong number.")
else:
    print(f"{n} is not an Armstrong number.")
```

```python
class Solution:
    def ranks(self, scoree):
        sorted_scores = sorted(scoree, reverse=True)
        rank_map ={}
        for i, score in enumerate(sorted_scores):
            if i == 0:
                rank_map[score] = "Gold Medal"
            elif i == 1:
                rank_map[score] = "silver Medal"
            elif i == 2:
                rank_map[score] = "Bronze Medal"
            else:
                rank_map[score] = str(i + 1)
        answered = [rank_map[score] for score in scoree]
        return answered
sol = Solution()
scoree = input("enter scores (comma-separated): ")
scoree = [int(x.strip()) for x in scoree.split(',')]
result = sol.ranks(scoree)
print("Ranks:", result)
    
```

```python
class Solution(object):
    def findRelativeRanks(self, score):
        """
        :type score: List[int]
        :rtype: List[str]
        """
        n = len(score)

        ranked = sorted(range(n), key = lambda i: -score[i])
        medals= ["Gold Medal", "Silver Medal", "Bronze Medal"]
        answer = [""] * n

        for rank,idx in enumerate(ranked):
            if rank < 3:
                answer[idx] = medals[rank]
            else:
                answer[idx] = str(rank + 1)

        return answer
sol = Solution()
score = input("Enter scores (comma-separated): ")
score = [int(x.strip()) for x in score.split(',')]
result = sol.findRelativeRanks(score)
print("Ranks:", result)
```
